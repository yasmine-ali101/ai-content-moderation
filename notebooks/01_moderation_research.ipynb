#Three Stage : IbrahimAmin --> Akhooli --> Camel Bert

```python
from transformers import pipeline, AutoTokenizer, AutoModel
import torch

# Load models
hate_clf = pipeline("text-classification", model="IbrahimAmin/marbertv2-finetuned-egyptian-hate-speech-classification")
toxicity_clf = pipeline("text-classification", model="akhooli/xlm-r-large-arabic-toxic")

# CAMeLBERT for final filter
tokenizer = AutoTokenizer.from_pretrained("CAMeL-Lab/bert-base-arabic-camelbert-mix")
camel_model = AutoModel.from_pretrained("CAMeL-Lab/bert-base-arabic-camelbert-mix")

def camel_filter(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True)
    with torch.no_grad():
        outputs = camel_model(**inputs)
    # Example heuristic: embedding norm as "suspicion score"
    norm_val = torch.norm(outputs.last_hidden_state.mean(dim=1)).item()
    return {"label": "ok"} if norm_val > 10 else {"label": "flagged"}

def moderate(text):
    # Stage 1: Hate Speech
    hate_result = hate_clf(text)[0]
    if hate_result["label"] != "Neutral":
        return {"stage": "hate_clf", "output": hate_result}

    # Stage 2: Toxicity
    tox_result = toxicity_clf(text)[0]
    if tox_result["label"] == "label_1":  # toxic
        return {"stage": "toxicity_clf", "output": tox_result}

    # Stage 3: Only if label_0 (non-toxic)
    if tox_result["label"] == 'LABEL_0':
        camel_result = camel_filter(text)
        return {"stage": "camelbert_filter", "output": camel_result}

    return {"stage": "final", "output": "allowed"}

# Example test cases
examples = [
    "إنت حمار وما بتفهمش حاجة.",      # Offensive
    "إنت واحد تافه ومحدش طايقك.",     # Offensive
    "الستات ما ينفعوش يشتغلوا.",      # Sexism
    "الناس من البلد دي كلهم كذابين.", # Racism
    "دينك كاذب",                       # Religious Discrimination
    "لازم نكسر ونولّع فيهم",             # Extremism
    "أنا عايز أقتل أي حد يخالفني.",   # Extremism
    "إنت شكلك شبه القرد.",             # Offensive
    "إيه اللبس المقرف ده يا معفّن؟",   # Offensive
    "اللي مش معانا يبقى عدو ولازم يتصفّى.", # Extremism
    "ياريت الناس دي كلها تختفي من الدنيا.", # Extremism
    "انتي غبية",                           # Offensive
    "إنت واحد محدش طايقك.",     # Offensive
    "أنا حزين",
    "اللي رأيه غير رأيي يبقى عدوى.",
    " اللي رأيه غير رأيي يبقى عدوى ولازم يموت"
    ]

for txt in examples:
    print("\nInput:", txt)
    print("Result:", moderate(txt))
```

#Final Pipeline:

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline

model_path = "khaled-kassem/Arabic-Profanity-Checker-v3"
model = AutoModelForSequenceClassification.from_pretrained(model_path)
tokenizer = AutoTokenizer.from_pretrained(model_path)

nlp_pipeline = pipeline("text-classification", model=model, tokenizer=tokenizer)
```

```python
# Load models
hate_clf = pipeline("text-classification", model="IbrahimAmin/marbertv2-finetuned-egyptian-hate-speech-classification")
toxicity_clf = pipeline("text-classification", model="akhooli/xlm-r-large-arabic-toxic")

def classify_text(text):
    # Step 1: Hate classifier
    hate_result = hate_clf(text)[0]  # result like {'label': 'neutral', 'score': ...

    if hate_result['label'].lower() == "neutral":
        # Step 2: Toxicity classifier if neutral
        tox_result = toxicity_clf(text)[0]
        if tox_result['label'] == "LABEL_1":
          tox_result['label'] = "Extremism"
        elif tox_result['label'] == "LABEL_0":
          tox_result['label'] = "Neutral"
        return {"stage": "toxicity", "result": tox_result}
    else:
        # Return hate result directly
        return {"stage": "hate", "result": hate_result}


for ex in examples:
    print(ex)
    print(classify_text(ex))
```

```python
def mask_profanity(sentence):
    words = sentence.split()
    masked_sentence = []

    for word in words:
        result = nlp_pipeline(word)[0]  # single word
        if result['label'] == 'LABEL_1':  # offensive
            masked_sentence.append("**")
        else:
            masked_sentence.append(word)

    return " ".join(masked_sentence)
```

```python
def two_model_pipeline(text):
    # Step 1: Main classifier
    result = classify_text(text)['result'] # Assuming classifier_text returns a list of dicts
    label = result['label']

    # Step 2: Generate explanation for detected category
    explanations = {
        "Racism": "هذا الكلام يحتوي على تمييز عنصري، وهذا غير مقبول لأنه يقلل من قيمة الناس بناءً على عرقهم أو جنسيتهم.",
        "Sexism": "هذا الكلام يتضمن تمييزًا على أساس الجنس، وهو غير مقبول لأن الجميع يستحقون المساواة والاحترام.",
        "Religious Discrimination": "هذا الكلام يسيء إلى معتقدات دينية، وهذا غير مقبول لأن حرية المعتقد حق للجميع.",
        "Offensive": "هذا الكلام مسيء وقد يجرح مشاعر الآخرين. استخدام لغة محترمة يساعد في التواصل بشكل أفضل.",
        "Extremism": "هذا الكلام يتضمن أفكارًا متطرفة، وهذا غير مقبول لأنه يشجع على العنف وعدم التسامح."
    }

    explanation = explanations.get(label, "هذا الكلام  مناسب ويجب تجنب استخدامه.")

    # Step 3: Check if offensive and mask toxic words
    categories_to_mask = ["Racism", "Offensive", "Extremism"]

    if label in categories_to_mask:
        masked_text = mask_profanity(text)
        return f"🚫 {explanation}\nالجملة المعدلة: {masked_text} \n"
    elif label not in categories_to_mask and label != "Neutral":
        return f"⚠ {explanation}\nالجملة الأصلية: {text} \n"
    else:
      return f"✅ الجملة سليمة ولا تحتوي على إساءة\nالجملة الأصلية: {text} \n"
```
